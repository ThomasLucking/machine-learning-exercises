```python
import pandas as pd
import joblib
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
# read the data so it is the correct dataframe.
df = pd.read_csv('../mission_1/data/titanic_cleaned.csv')

# we define Y as the main goal so survived
Y = df['Survived']
# x for everything else
X = df.drop(columns=['Survived'])

# then we split the data into 80% for training and 20% for tests
X_train, X_test, y_train, y_test = train_test_split(
    X, Y, test_size=0.2, train_size=0.8
)

Lr = LogisticRegression(random_state=0, max_iter=1000).fit(X, Y) # updated the max amount of iterations.

joblib.dump(Lr, 'logic_regression_mode.joblib')
```
